# Bogotá

```python
df_bogota=pd.read_excel("DatasetRaw/2Lab02_anexo-proyecciones-poblacion-bogota_UPZ-2018-2024.xlsx", skiprows=11)
```

```python
df_bogota
```

```python
df_bogota = df_bogota.groupby(['NOM_LOC', 'AÑO']).agg({
    'COD_LOC': 'first',  # Conservar el primer valor de COD_LOC por grupo
    'TOTAL': 'sum'       # Sumar los valores de TOTAL por grupo
}).reset_index()
```

```python
df_bogota[df_bogota["AÑO"]<2031]
```

```python
df_pivot = df_bogota.pivot_table(
    index=['NOM_LOC', 'COD_LOC'],  # Las columnas que queremos mantener como índice (BARRIO y COD_LOC)
    columns='AÑO',                 # Columna a pivotar (años)
    values='TOTAL',                # Columna que contiene los valores (población total)
    aggfunc='sum'                  # En caso de que haya duplicados, usar la suma (aunque debería ser único)
).reset_index()

# Renombrar la columna 'NOM_LOC' a 'BARRIO'
df_pivot.rename(columns={'NOM_LOC': 'BARRIO'}, inplace=True)
```

```python
df_pivot["Ciudad"]="Bogotá"
```

```python
df_pivot
```

Incluyamos las coordenadas de Bogota

```python
file_path = 'DatasetRaw/3LabLocBogota.csv'
df_lab_loc = pd.read_csv(file_path, delimiter=";")
df_lab_loc 
```

```python
df_pivot['COD_LOC'] = df_pivot['COD_LOC'].astype(int)
df_lab_loc['CODIGO'] = df_lab_loc['CODIGO'].astype(int)
```

```python

# Realizar el merge (join) entre ambos DataFrames por la columna COD_LOC y CODIGO
df_merged = pd.merge(df_pivot, df_lab_loc, left_on='COD_LOC', right_on='CODIGO', how='left')

# Guardar el resultado en un archivo CSV o mostrar el resultado
df_merged.to_csv('resultado_merge.csv', index=False)

# O también puedes mostrar las primeras filas
```
